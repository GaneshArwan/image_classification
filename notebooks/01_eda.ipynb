# 01. Exploratory Data Analysis — Plant Disease Classification

Inspect class distribution, sample leaf images, pixel statistics, and augmentation transforms.

```
import sys
from pathlib import Path
from collections import Counter
import matplotlib.pyplot as plt
from torchvision import datasets

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.data import build_transforms, generate_synthetic_leaf_dataset
```

```
data_dir = ROOT / "data/plantvillage"
if not any(p.is_dir() for p in data_dir.glob("*")) if data_dir.exists() else True:
    data_dir = generate_synthetic_leaf_dataset(ROOT / "artifacts/notebook_data", image_size=128)
dataset = datasets.ImageFolder(str(data_dir))
counts = Counter(dataset.targets)
print("Classes:", dataset.classes)
print("Counts:", {dataset.classes[k]: v for k, v in counts.items()})
```

```
train_tf, _ = build_transforms(image_size=128, augment=True)
fig, axes = plt.subplots(2, min(3, len(dataset.classes)), figsize=(10, 6))
for idx, cls_name in enumerate(dataset.classes[:3]):
    sample_idx = dataset.targets.index(idx)
    img, _ = dataset[sample_idx]
    axes[0, idx].imshow(img)
    axes[0, idx].set_title(f"{cls_name} (raw)")
    axes[0, idx].axis("off")
    aug_tensor = train_tf(img)
    aug_vis = (aug_tensor.permute(1, 2, 0).numpy() * 0.225 + 0.485).clip(0, 1)
    axes[1, idx].imshow(aug_vis)
    axes[1, idx].set_title(f"{cls_name} (augmented)")
    axes[1, idx].axis("off")
plt.tight_layout()
```
