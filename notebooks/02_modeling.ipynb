# 02. Modeling, Augmentation Comparison, and Grad-CAM

Compare Baseline CNN vs. ResNet18 with and without data augmentation, then inspect confusion matrices and Grad-CAM.

```
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.data import generate_synthetic_leaf_dataset
from src.i18n import Translator
from src.schemas import TrainConfig
from src.train import train_single_run
```

```
data_dir = generate_synthetic_leaf_dataset(ROOT / "artifacts/notebook_data", image_size=64)
tr = Translator("en")
results = []
for arch, aug in [("baseline_cnn", False), ("baseline_cnn", True), ("resnet18", False), ("resnet18", True)]:
    cfg = TrainConfig(data_dir=data_dir, output_dir=ROOT / "artifacts/notebook_runs", architecture=arch, epochs=2, batch_size=8, image_size=64, use_augmentation=aug, pretrained=False)
    results.append(train_single_run(cfg, tr))
```

```
for r in results:
    print(f"{r.architecture:14s} aug={str(r.augmented):5s} acc={r.accuracy:.3f} f1={r.macro_f1:.3f} auc={r.macro_roc_auc}")
```
